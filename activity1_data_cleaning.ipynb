```python
#calculates outliers
numeric_cols = df.select_dtypes(include=np.number).columns

Q1 = df[numeric_cols].quantile(0.25)
Q3 = df[numeric_cols].quantile(0.75)

IQR = Q3 - Q1

outliers = (
    (df[numeric_cols] < Q1 - 1.5 * IQR) |
    (df[numeric_cols] > Q3 + 1.5 * IQR)
)

print("Outliers:")
print(outliers.sum())
```

```python
# Check column names
print(df.columns)

# Check data types
print(df.dtypes)

# Check unique values in text columns
for col in df.select_dtypes(include="object").columns:
    print(col, df[col].unique()[:10])
```

```python
#remove unwanted rows
df = df.dropna(how="all")
print("Unwanted rows removed.")
```

```python
#remove duplicates rows 
df = df.drop_duplicates()
print("Duplicate records removed.")
```

```python
# Clean column names
df.columns = df.columns.str.strip()

# Clean text values
print("Inconsistent spaces corrected.")
```

```python
# Numerical columns
numeric_cols = df.select_dtypes(include=np.number).columns

# Fill missing numerical values with median
for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())

# Categorical columns
categorical_cols = df.select_dtypes(include="object").columns

# Fill missing categorical values with mode
for col in categorical_cols:
    if not df[col].mode().empty:
        df[col] = df[col].fillna(df[col].mode()[0])

print("Missing values handled.")
```

```python
for col in numeric_cols:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df[col] = df[col].clip(lower, upper)

print("Outliers handled using IQR.")
```

```python
#check pandas execution time
import time
start_time = time.time()
# Load dataset
pandas_df = pd.read_csv("data.csv")
# Remove empty rows and duplicates
pandas_df = pandas_df.dropna(how="all")
pandas_df = pandas_df.drop_duplicates()
# Handle missing numerical values
num_cols = pandas_df.select_dtypes(include=np.number).columns
for col in num_cols:
    pandas_df[col] = pandas_df[col].fillna(
        pandas_df[col].median()
    )
# Handle missing categorical values
cat_cols = pandas_df.select_dtypes(include="object").columns
for col in cat_cols:
    if not pandas_df[col].mode().empty:
        pandas_df[col] = pandas_df[col].fillna(
            pandas_df[col].mode()[0]
        )
# Measure execution time
pandas_time = time.time() - start_time
print("Pandas Execution Time:", pandas_time, "seconds")
```
